# VAMOS Notebook Index

This index tracks the maintained notebook set. The learning path is organized by tier and aligned with the current public API.

## 0. Basic Tier

- [00_api_comparison.ipynb](0_basic/00_api_comparison.ipynb): compare the unified API, explicit config objects, and CLI-driven runs.
- [01_quickstart.ipynb](0_basic/01_quickstart.ipynb): first optimization run with `optimize(...)`.
- [02_visualizing_results.ipynb](0_basic/02_visualizing_results.ipynb): plotting, tables, and result inspection.
- [03_compare_algorithms_quick.ipynb](0_basic/03_compare_algorithms_quick.ipynb): quick side-by-side algorithm comparison.
- [04_advanced_configuration.ipynb](0_basic/04_advanced_configuration.ipynb): explicit config builders and external archives.
- [05_interactive_tutorial.ipynb](0_basic/05_interactive_tutorial.ipynb): guided end-to-end walkthrough.
- [06_optuna_tuning_basics.ipynb](0_basic/06_optuna_tuning_basics.ipynb): introductory Optuna-based tuning workflow.
